# Run simple simulation loaded from APIs

```python
import PyRAMSES
```

```python
ram = PyRAMSES.sim()
```

```python
ram.addIOFile('dyn_A.dat','DAT')
ram.addIOFile('volt_rat_A.dat','DAT')
ram.addIOFile('settings1.dat','DAT')
ram.addIOFile('init.trace','INIT')
ram.addIOFile('short_trip_branch.dst','DST')
ram.addIOFile('cont.trace','CONT')
ram.addIOFile('disc.trace','DISC')
```

```python
ram.initSim()
```

```python
ram.initPF()
```

```python
ram.initDyn()
```

```python
ram.initVars()
```

```python
ram.addIOFile('BV g1','RTOBS')
```

```python
ram.execSim(10.0)
```

```python
ram.contSim(150.0)
```

```python
ram.endSim()
```

```python
del(ram)
```
